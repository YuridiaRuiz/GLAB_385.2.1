#### Data Collections
- lists (arrays) - indexed collection. Have order and stay in order. Number by index
- dictionaries (dict) - objects - Keyed collection, Have no order, use key to access information.
- tuples -records (rows of a table)

### Strings

    - Collection of single characteres

```python
greet = 'Hello, Wrold!'

#print the first value in a string/list
print(greet[0])
print(greet[1])
```

### Python Indexing operator
- can be used with strings and list
- `[start : stop : step]`
- by default, if we don't include the step,its 1

```python
print(greet[0:5:1])
print(greet[0::2]) #print letters each 2 letters,
print(greet[::-1]) #print reverse start in the last index, goes backwards
```

### Type() Function
- tell us data type a variable is
- `type(value) -> data type`

```python
print(type(greet))
print(type(13))
print(type(True))
print(type(13.6))
```

### Typecasting Functions (to change the datatype of)
- `str()` - turns anything into a string
- `float()` - turns value into a float if possible
- `int()` - turns calue into a integer if possible

```python
age = '30'
print(type(age))

#If you permanently want to chanfe a variable you have to re-assign it
age = int(age)
print(type(age))
```

#### Methods - built in function to python objects (everything in python is technically an object), for different datatypes/objects/modules
