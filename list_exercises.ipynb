# Reference
- creating a copy of a list DOES NOT create a new list.It points to the same place in memory
- Google: Pass by value vs Pass by Reference
- Changing one referenced list, will also change the other.

```python
og_list = [1,2,3,4,5]

new_list = og_list

print('OG',og_list)
print('New',new_list)

new_list.append(6)

print('OG',og_list)
print('New',new_list)
```

```python
# Deep copy -  a complety separate copy og a Non-primitive data type, not just a reference to the same place in memory
# Shallow copy - a semi separate copy of a Non-primitive data type. 1st dimension is separate places in memory
second_copy = og_list.copy()

print('OG',og_list)
print('New',new_list)
print('Second copy:', second_copy)

og_list.append(7)

print('OG',og_list)
print('New',new_list)
print('Second copy:', second_copy)
```

```python
#Reverse, extends, sort
num_list = [1,2,3,4,5]
print(num_list)

num_list.reverse()
print(num_list)

num_list.extend([6,7,8,9])
print(num_list)

num_list.sort()
print(num_list)
```

```python
#Membership operators - useful for conditionals
print(1 in num_list)
print(4 not in num_list)
```

```python
# Creat a function,take in 2 list of year, reverse so its in chronological order, add the, togheter
# Create reusable codeblock, can be reused as many times as needed
# Can take input, AND can provide output

# Defined a function called 'reorder_profit_years' 
# Takes 2 parameters (year1, year2)
def reorder_profit_years(years1, years2):
    # Reversed the order of years for both list
    years1.reverse()
    years2.reverse()

    # Added list together
    years1.extend(years2)

    #return the newly extended list
    return years1
```

```python
year_list1 = [2026,2025,2024,2023]
year_list2 = [1990,1989,1988,1987]

print(reorder_profit_years(year_list2,year_list1))
```
